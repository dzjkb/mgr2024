```python
import pandas as pd

from src.ds_utils import load_audio_tensor, AudioTensor
```

```python
encoded = load_audio_tensor("/media/Data/jp_dir/sounds/vctk_words_v2/audio_val_in_latent_space")
labels = pd.read_csv("/media/Data/jp_dir/sounds/vctk_words_v2/labels.csv")

fnames = '\n'.join(encoded.filenames[:10])
print(f"got {len(encoded.data)} encoded samples, with filenames like:\n{fnames}")
```

```python
def _get_labels(enc: AudioTensor, label_column: str) -> dict[int, str]:
    idx_to_speaker = {idx: spk.split("_")[0] for idx, spk in enumerate(enc.filenames)}
    speaker_to_label = dict(labels[["speaker_id", label_column]].set_index("speaker_id", drop=True)[label_column].items())

    return {
        idx: speaker_to_label[spk]
        for idx, spk in idx_to_speaker.items()
    }
```

```python
gender_labels = _get_labels(encoded, "GENDER")
accent_labels = _get_labels(encoded, "ACCENTS")
```
